```python
import torch
from diffusers import QwenImage21Pipeline

pipe = QwenImage21Pipeline.from_pretrained(
    "Qwen/Qwen-Image-2.1", torch_dtype=torch.bfloat16
)
```

```python
print(pipe.transformer)
```

`pos_embed: QwenImage21Rope()` — Applies rotary positional embeddings to attention Q/K so the model knows spatial/sequence position of tokens.

`time_text_embed: QwenImage21TimestepProjEmbeddings()` — Converts the diffusion/flow timestep into a 4096-D conditioning vector used throughout the transformer.

`time_proj: QwenImage21TemporalTimesteps()` — Encodes scalar timestep `t` into a 256-D sinusoidal-style representation.

`timestep_embedder.linear_1: Linear(256, 4096)` — Projects the 256-D timestep representation into the model hidden size.

`timestep_embedder.act: SiLU()` — Adds nonlinearity to the timestep embedding.

`timestep_embedder.linear_2: Linear(4096, 4096)` — Produces the final 4096-D timestep embedding.

`txt_in: QwenImage21TextProjection()` — Adapts Qwen text/context embeddings into the diffusion transformer representation space.

`txt_in.text_norm: QwenImage21ZeroCenterRMSNorm()` — Normalizes incoming text/context embeddings before projection.

`txt_in.in_layer: Linear(4096, 4096)` — First learned projection of text/context features.

`txt_in.act: GELU()` — Nonlinear activation in the text projection MLP.

`txt_in.out_layer: Linear(4096, 4096)` — Produces final 4096-D text/context tokens for the transformer.

`img_in: Linear(64, 4096)` — Projects each 64-channel VAE latent token into the 4096-D transformer hidden space.

`modulation: Sequential(SiLU, Linear(4096, 16384))` — Converts timestep conditioning into modulation vectors used to scale/shift/gate transformer computations.

`modulation.0: SiLU()` — Nonlinear activation before producing modulation parameters.

`modulation.1: Linear(4096, 16384)` — Expands the timestep embedding into four 4096-D chunks of modulation information.

`transformer_blocks: ModuleList(32 × QwenImage21TransformerBlock)` — The main stack of 32 transformer blocks that jointly process context and image tokens.

`img_norm1: LayerNorm(4096, affine=False)` — Normalizes hidden states before attention, with adaptive modulation supplied externally instead of learned LayerNorm affine parameters.

`attn: QwenImage21Attention()` — Performs multi-head attention between the tokens in the joint sequence.

`attn.to_q: Linear(4096, 4096)` — Projects hidden states into Query vectors.

`attn.to_k: Linear(4096, 4096)` — Projects hidden states into Key vectors.

`attn.to_v: Linear(4096, 4096)` — Projects hidden states into Value vectors.

`attn.norm_q: RMSNorm()` — Normalizes Query vectors before computing attention logits.

`attn.norm_k: RMSNorm()` — Normalizes Key vectors before computing attention logits.

`attn.to_out.0: Linear(4096, 4096)` — Recombines the multi-head attention output back into the model hidden space.

`attn.to_out.1: Dropout(p=0.0)` — Optional dropout after attention output projection; disabled here because `p=0`.

`img_norm2: LayerNorm(4096, affine=False)` — Normalizes hidden states before the feed-forward/SwiGLU sublayer.

`img_mlp: QwenImage21SwiGLUFeedForward()` — Per-token nonlinear feed-forward network used after attention.

`img_mlp.proj: Linear(4096, 12288)` — Expands each token from 4096 to 12288 features for the main SwiGLU branch.

`img_mlp.gate_layer: Linear(4096, 12288)` — Creates the parallel gate branch used by SwiGLU.

`img_mlp.activation_fn: SiLU()` — Activates the gate branch before elementwise multiplication.

`img_mlp.out: Linear(12288, 4096)` — Projects the gated 12288-D MLP representation back to 4096 dimensions.

`norm_out: QwenImage21AdaLayerNormContinuous()` — Final timestep-conditioned normalization applied after the 32 transformer blocks.

`norm_out.silu: SiLU()` — Nonlinearity used to process the conditioning vector for the final adaptive normalization.

`norm_out.linear: Linear(4096, 4096)` — Projects conditioning information used by the final adaptive LayerNorm.

`norm_out.norm: LayerNorm(4096, affine=False)` — Final normalization of transformer hidden states before output projection.

`proj_out: Linear(4096, 64)` — Converts transformer hidden states back from 4096 dimensions to the 64-channel VAE latent/flow prediction space.
